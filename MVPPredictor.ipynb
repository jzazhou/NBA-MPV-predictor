#Downsampling  , making the 'No' class smaller

```python
MVPdataset = df.iloc[:, 3:13]
```

```python
MVPArray = MVPdataset.to_numpy()
MVPArray.astype(float)
```

```python
print(len(MVPArray))
print(MVPArray)
```

# Features dataset

```python
import numpy as np
```

```python
features_train = np.concatenate((MVPArray,NonMVPArray))
print(len(features_train))
```

# Targets dataset

```python
target_train = []
for x in range(22):
  target_train.append('Yes')
for i in range(22):
  target_train.append('No')
print(len(target_train))
```

# Training with Decision Tree Classifier

```python
from sklearn.model_selection import train_test_split
features_train,features_test,target_train,target_test=train_test_split(features_train, target_train, test_size=0.2, random_state=123)
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(features_train, target_train)
```

```python
features_train
```

```python
expected = target_test
predicted = model.predict(features_test)
```

```python
expected
```

```python
predicted
```

```python
from sklearn import metrics
print(metrics.classification_report(expected, predicted))
print(metrics.confusion_matrix(expected, predicted))
```

#Upsampling with SMOTE

```python
from imblearn.over_sampling import SMOTE
```

```python
MVP_complete = df.iloc[:,3:13]
MVPArray2 = MVP_complete.to_numpy()
MVPArray2.astype(float)
Non_MVP_complete = df2.iloc[:,3:13]
NonMVPArray2 = Non_MVP_complete.to_numpy()
NonMVPArray2.astype(float)
X_train = np.concatenate((MVPArray2,NonMVPArray2,), axis=0)
```

```python
y_train = []
for x in range(len(MVPArray2)):
  y_train.append('Yes')
for i in range(len(NonMVPArray2)):
  y_train.append('No')
print(len(y_train))
```

```python
X_train, X_test, y_train, y_test=train_test_split(X_train, y_train, test_size=0.3, random_state=42)

sm=SMOTE(random_state=123)

X_train_res, y_train_res= sm.fit_resample(X_train, y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
```

```python
from sklearn.model_selection import train_test_split
features_train,features_test,target_train,target_test=train_test_split(X_train_res, y_train_res, test_size=0.2, random_state=123)
```

```python
len(features_train)
```

```python
model.fit(features_train, target_train)
```

```python
expected = target_test
predicted = model.predict(features_test)
```

```python
from sklearn import metrics
print(metrics.classification_report(expected, predicted))
print(metrics.confusion_matrix(expected, predicted))
```
